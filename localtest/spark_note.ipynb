```python
import findspark
findspark.init()

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("LocalNotebookApp1") \
    .getOrCreate()

spark
datos = [
    ("1", "Juan", "Quito"),
    ("2", "Ana", "Guayaquil"),
    ("3", "Luis", "Cuenca")
]

columnas = ["id", "nombre", "ciudad"]

df = spark.createDataFrame(datos, columnas)

# ✅ Mostrar contenido
df.show()
```
